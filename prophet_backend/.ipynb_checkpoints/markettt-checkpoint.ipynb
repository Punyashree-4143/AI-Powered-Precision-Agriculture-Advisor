```python
import pandas as pd

# Load all CSVs
import glob

files = glob.glob(r"C:\Users\punya\Downloads\market\csv\*.csv")

df_list = []
for f in files:
    print("Loading:", f)
    df_list.append(pd.read_csv(f))

df = pd.concat(df_list, ignore_index=True)
```

```python
df = df[df['State'] == 'Karnataka'].copy()

df['Arrival_Date'] = pd.to_datetime(df['Arrival_Date'], errors='coerce')
df = df.dropna(subset=['Arrival_Date', 'Modal_Price'])

df = df.sort_values('Arrival_Date')
```

```python
def create_features(ts_df, lags=[1,7,14], roll_windows=[7,14]):
    df_feat = ts_df.copy()
    df_feat = df_feat.sort_values("Arrival_Date")

    for l in lags:
        df_feat[f"lag_{l}"] = df_feat["Modal_Price"].shift(l)

    for w in roll_windows:
        df_feat[f"roll_mean_{w}"] = df_feat["Modal_Price"].shift(1).rolling(w).mean()

    df_feat = df_feat.dropna()
    return df_feat
```

```python
from xgboost import XGBRegressor
import joblib
import os

os.makedirs("models", exist_ok=True)

grouped = df.groupby(["District", "Market", "Commodity"])

for (district, market, commodity), gdf in grouped:
    print(f"Training → {district} / {market} / {commodity}")

    ts = gdf[["Arrival_Date", "Modal_Price"]].reset_index(drop=True)

    # Create features
    ts_feat = create_features(ts)

    if len(ts_feat) < 200:
        print("➡ Skipped (not enough data)\n")
        continue

    X = ts_feat.drop(["Arrival_Date", "Modal_Price"], axis=1)
    y = ts_feat["Modal_Price"]

    # Train-Test Split
    split_idx = int(len(X) * 0.80)
    X_train, X_test = X.iloc[:split_idx], X.iloc[split_idx:]
    y_train, y_test = y.iloc[:split_idx], y.iloc[split_idx:]

    # Train XGBoost model
    model = XGBRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        objective='reg:squarederror'
    )

    model.fit(X_train, y_train)

    # Save model
    safe_name = f"{district}_{market}_{commodity}".replace(" ", "_").replace("/", "_")
    joblib.dump(model, f"models/{safe_name}.joblib")

    print(f"✔ Saved: models/{safe_name}.joblib\n")
```

```python
import os

models = [f for f in os.listdir("models") if f.endswith(".joblib")]

commodities = set()

for m in models:
    # Remove .joblib
    name = m.replace(".joblib", "")

    # Split by "_"
    parts = name.split("_")

    # Commodity name is everything AFTER District + Market
    # District and Market may have multiple parts → so we extract the last part(s)
    # But commodity may contain underscores, so we join everything after 2nd part
    commodity = "_".join(parts[2:])

    commodities.add(commodity)

print("Total unique commodities trained:", len(commodities))
for c in sorted(commodities):
    print(c)
```

```python
import pandas as pd
from datetime import datetime, timedelta

def forecast_7_days(model, last_ts_df, days=7):
    future_preds = []
    temp = last_ts_df.copy()

    # Start forecasting from today's real date
    start_date = pd.to_datetime(datetime.today().date())

    for i in range(days):
        # Create features from the latest available data
        feat = create_features(temp).iloc[-1:].drop(["Arrival_Date", "Modal_Price"], axis=1)

        next_price = model.predict(feat)[0]

        # Real future date from TODAY
        next_date = start_date + timedelta(days=i)

        # Add synthetic predicted point into the series for next iteration
        temp = pd.concat([
            temp,
            pd.DataFrame({"Arrival_Date":[next_date], "Modal_Price":[next_price]})
        ], ignore_index=True)

        future_preds.append({
            "date": next_date.strftime('%Y-%m-%d'),
            "predicted_price": float(next_price)
        })

    return future_preds
```

```python
import os

def load_model(district, market, commodity):
    safe_d = district.replace(" ", "_").replace("/", "_")
    safe_m = market.replace(" ", "_").replace("/", "_")
    safe_c = commodity.replace(" ", "_").replace("/", "_")

    for f in os.listdir("models"):
        if f.startswith(f"{safe_d}_{safe_m}_{safe_c}"):
            return joblib.load(os.path.join("models", f))
    return None
```

```python
model = load_model("Dharwad", "Hubli (Amaragol)", "Groundnut")

if model is None:
    print("Model not found!")
else:
    print("Model loaded successfully.")
```

```python
import os
import joblib

def load_model(district, market, commodity):
    safe_d = district.replace(" ", "_").replace("/", "_")
    safe_m = market.replace(" ", "_").replace("/", "_")
    safe_c = commodity.replace(" ", "_").replace("/", "_")

    for f in os.listdir("models"):
        if f.startswith(f"{safe_d}_{safe_m}_{safe_c}"):
            return joblib.load(os.path.join("models", f)), f
    
    return None, None
```

```python
def test_prediction(district, market, commodity, days=7):
    model, filename = load_model(district, market, commodity)

    if model is None:
        print("❌ Model not found for:", district, market, commodity)
        return
    
    print("✔ Loaded model:", filename)

    ts = df[(df["District"] == district) &
            (df["Market"] == market) &
            (df["Commodity"] == commodity)][["Arrival_Date","Modal_Price"]]

    preds = forecast_7_days(model, ts, days)
    
    for p in preds:
        print(p)
```

```python
test_prediction("Belgaum", "Ramdurga", "Groundnut")
```

```python
test_prediction("Belgaum", "Nippani", "Rice")
```
